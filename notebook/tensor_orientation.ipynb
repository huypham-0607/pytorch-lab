```python
# First tensor

data = [[17,70],[1,3]]
x_data = torch.tensor(data)

print(x_data)

np_array = np.array(data)
print(type(np_array))
x_np = torch.tensor(np_array)
print(x_data)
print(type(x_data))
```

```python
"""
Initializing with another tensor
"""

# Inherit shapes and dtype from x_data
x_ones = torch.ones_like(x_data)
print(x_ones)

# Same with first but with zeros
x_zeros = torch.zeros_like(x_ones)
print(x_zeros)

# Create tensor with shape of x_data, but overriding dtype
x_rand_2 = torch.rand_like(x_data, dtype=torch.float)
print(x_rand_2)
```

```python
"""
    Initializing with random or constant value
"""

shape = (2,3)

# Create new tensor prefilled with ones, zeroes, and random
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)
rand_tensor = torch.rand(shape)

print(ones_tensor)
print(zeros_tensor)
print(rand_tensor)
```

```python
"""
    Attributes of a tensor
"""

tensor = torch.rand(3,4)

print(f"Shape: {tensor.shape}")
print(f"Datatype: {tensor.dtype}")
print(f"Device tensor is stored on: {tensor.device}")
```

```python
"""
    Tensor operations can be run on CPU and Accelerator (CUDA, MPS, MTIA, XPU).

    By default, tensors are created on CPU. Use the tensor.to() method to move tensor to Accelerator
    (After checking their availability).

    Keep in mind copying tensor across devices can be expensive in time and memory
"""

print(torch.accelerator.current_accelerator())

tensor = torch.rand(4,4)

if torch.accelerator.is_available():
    tensor = tensor.to(torch.accelerator.current_accelerator())

print(tensor)
```

```python
"""
    Some slicing/indexing tensor operations

    These are very NumPy adjacent syntax
"""

tensor = torch.rand(4,4)

# Slicing and indexing
print(f"First row: {tensor[0]}")
print(f"First column: {tensor[:,0]}")
print(f"Last column: {tensor[:,-1]}")

# Concatenating tensor along a dimension
print(torch.cat([tensor,tensor],dim=1))
print(torch.cat([tensor,tensor],dim=0))
```

```python
"""
    Matrix multiplication
"""

tensor = torch.rand(3,4)

print(f"OG tensor:\n {tensor}")

# All of these describes standard matrix multiplication
y1 = tensor @ tensor.T
y2 = tensor.matmul(tensor.T)

y3 = torch.empty(0)
torch.matmul(tensor, tensor.T, out=y3)

print(y1)
print(y2)
print(y3)
```

```python
"""
    Element-wise product
"""

tensor = torch.rand(3,4)
print(f"OG tensor:\n{tensor}")

# Again, all three methods achieve the same element-wise multiplication
z1 = tensor * tensor
z2 = tensor.mul(tensor)

z3 = torch.empty(0)
z3 = torch.mul(tensor,tensor,out=z3)
print(z1)
print(z2)
print(z3)
```

# Tensors

Mathematically, tensor is a generalization of a scalar, vector, or a matrix. They are characterized by their order, which specify the number of dimension a tensor can have. A scalar is a tensor with order 0, vector is a tensor with order 1, matrix is a tensor with order 2, etc. In a computational sense, we can think of tensors as a multi-dimensional array.

In PyTorch, tensor is implemented as an efficient data container supporting a multitude of operations. It is somewhat similar to a numpy array, though it supports more features useful to deep learning (differentiation, etc), and can run on a GPU.
